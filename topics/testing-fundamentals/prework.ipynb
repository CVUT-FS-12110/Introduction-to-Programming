## Úkol 1: Napište testy pro funkci `calculate_grade` (10 min)

Následující funkce převádí bodové skóre na známku.
Napište **alespoň 6 testů** — normální případy i edge cases.

Nezapomeňte otestovat:
- Typické hodnoty z každého pásma
- Hraniční hodnoty (přesně 90, přesně 0, přesně 100)
- Neplatné vstupy (záporné číslo, víc než 100)

````python
%%writefile grade_calculator.py
"""Grade calculator module."""


def calculate_grade(score: int) -> str:
    """Convert a numeric score (0-100) to a letter grade.
    
    Grading scale:
        90-100 -> 'A'
        80-89  -> 'B'
        70-79  -> 'C'
        60-69  -> 'D'
        0-59   -> 'F'
    
    Raises ValueError if score is not in range 0-100.
    """
    if not isinstance(score, int):
        raise ValueError("Score must be an integer")
    if score < 0 or score > 100:
        raise ValueError("Score must be between 0 and 100")
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    elif score >= 60:
        return 'D'
    else:
        return 'F'
````

### Inspirace (pokud nevíte, co vybrat)

**A) Validátor rodného čísla**

Napište funkci, která ověří, zda je české rodné číslo platné.
Formát: `YYMMDD/XXXX` (lomítko může i chybět). Pravidla:
ženy mají k měsíci přičteno 50 (leden ženy = 51), den musí existovat
v daném měsíci, rodná čísla po roce 1954 musí být dělitelná 11.
Přemýšlejte: co přestupný rok a 29. únor? Co měsíc 13 nebo den 32?
Co devítimístné RČ (před rokem 1954)? Co nuly na začátku?

**B) Validátor bankovního účtu a IBAN**

Napište funkce pro validaci českého čísla účtu i mezinárodního
formátu IBAN. České číslo má formát `předčíslí-číslo/kód_banky`
(např. `19-2000145399/0800`), kde předčíslí je nepovinné,
číslo účtu má max 10 číslic, kód banky 4 číslice. Obě části se
validují modulo 11 s váhami. IBAN (`CZ6508000000192000145399`)
se validuje přesunem prvních 4 znaků na konec, převodem písmen
na čísla a kontrolou modulo 97. Přemýšlejte: co když chybí
lomítko? Co samé nuly? Co IBAN s mezerami nebo malými písmeny?
Co neexistující kód banky? Co IBAN z jiné země (jiná délka)?

**C) Validátor a formátovač telefonního čísla**

Napište funkci, která přijme telefonní číslo v libovolném formátu
a vrátí ho v jednotném tvaru `+420 XXX XXX XXX`.
Musí zvládnout: `+420123456789`, `00420 123 456 789`,
`123-456-789`, `+421...` (slovenské — platné, nebo ne?),
příliš krátké/dlouhé, písmena ve vstupu, prázdný řetězec.
Přemýšlejte o všech způsobech, jak lidé reálně píšou telefonní čísla.

**D) Parser a validátor české poštovní adresy**

Napište funkci, která z řetězce extrahuje jednotlivé části
české adresy: ulice + číslo popisné/orientační, město, PSČ.
Formáty: `Thákurova 9, 160 00 Praha 6` nebo `Thákurova 2077/9`
(číslo popisné/orientační). PSČ musí být 5 číslic (s nebo bez mezery).
Přemýšlejte: co adresy bez ulice (malé obce)? Co číslo evidenční?
Co PSČ s neplatným rozsahem? Co chybějící čárky?

**E) Stahování a parsování kurzů ČNB z API**

Napište funkci, která stáhne aktuální kurzy ČNB z jejich
veřejného textového souboru a vrátí kurz zadané měny
(např. `get_rate("EUR")` → `25.120`).
Přemýšlejte: co když je API nedostupné? Co neexistující měna?
Co formát odpovědi — jak ho spolehlivě naparsovat?
AI se typicky sekne na parsování reálného formátu ČNB
a na ošetření síťových chyb.
