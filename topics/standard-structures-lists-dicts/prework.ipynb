## Les: spusť a prohlédni

Tohle je **les** — hotový funkční program „Tvůj rok v hudbě".
Tvým úkolem **není ho psát**. Spusť ho a dívej se, jak v něm spolupracují
**listy** (historie přehrání) a **slovníky** (jedno přehrání, počty interpretů).
Na cviku tenhle les rozebereme na malé **stromy** a složíš je sám/sama.

```python
%%writefile rok_v_hudbe.py
"""Tvuj rok v hudbe - LES (hotovy funkcni program). Nepis ho, spust ho."""

# Kazde prehrani = slovnik (dict). Cela historie = seznam (list) prehrani.
prehravani = [
    {"interpret": "Lucie",    "skladba": "Sen",        "delka_s": 214, "preskoceno": False, "datum": "2025-01-03"},
    {"interpret": "Kabat",    "skladba": "Pohoda",     "delka_s": 198, "preskoceno": True,  "datum": "2025-01-03"},
    {"interpret": "Lucie",    "skladba": "Amerika",    "delka_s": 251, "preskoceno": False, "datum": "2025-01-05"},
    {"interpret": "Chinaski", "skladba": "Vrat mi ji", "delka_s": 233, "preskoceno": False, "datum": "2025-01-05"},
    {"interpret": "Kabat",    "skladba": "Burlaci",    "delka_s": 205, "preskoceno": False, "datum": "2025-02-01"},
    {"interpret": "Lucie",    "skladba": "Sen",        "delka_s": 214, "preskoceno": False, "datum": "2025-02-01"},
    {"interpret": "Mig 21",   "skladba": "Snadne",     "delka_s": 187, "preskoceno": True,  "datum": "2025-02-14"},
    {"interpret": "Chinaski", "skladba": "Punk",       "delka_s": 176, "preskoceno": False, "datum": "2025-03-02"},
    {"interpret": "Lucie",    "skladba": "Amerika",    "delka_s": 251, "preskoceno": False, "datum": "2025-03-02"},
    {"interpret": "Kabat",    "skladba": "Pohoda",     "delka_s": 198, "preskoceno": False, "datum": "2025-03-10"},
    {"interpret": "Mig 21",   "skladba": "Hadr",       "delka_s": 192, "preskoceno": False, "datum": "2025-03-10"},
    {"interpret": "Lucie",    "skladba": "Medvidek",   "delka_s": 240, "preskoceno": True,  "datum": "2025-04-01"},
    {"interpret": "Chinaski", "skladba": "Punk",       "delka_s": 176, "preskoceno": False, "datum": "2025-04-01"},
    {"interpret": "Lucie",    "skladba": "Sen",        "delka_s": 214, "preskoceno": False, "datum": "2025-04-18"},
    {"interpret": "Kabat",    "skladba": "Burlaci",    "delka_s": 205, "preskoceno": False, "datum": "2025-05-09"},
    # >>> Sem pridej alespon 3 sva prehrani (zkopiruj radek, zmen hodnoty) <<<
]


def pocet_prehrani(prehravani):
    return len(prehravani)


def unikatni_interpreti(prehravani):
    jmena = set()
    for p in prehravani:
        jmena.add(p["interpret"])
    return jmena


def pocty_podle_interpreta(prehravani):
    pocty = {}
    for p in prehravani:
        pocty[p["interpret"]] = pocty.get(p["interpret"], 0) + 1
    return pocty


def top_interpreti(pocty, kolik=5):
    dvojice = list(pocty.items())                    # slovnik -> seznam dvojic
    dvojice.sort(key=lambda d: d[1], reverse=True)   # serad podle poctu
    return dvojice[:kolik]                           # ukroj prvnich N


def procento_preskocenych(prehravani):
    preskocene = [p for p in prehravani if p["preskoceno"]]   # comprehension = filtr
    if not prehravani:
        return 0.0
    return 100 * len(preskocene) / len(prehravani)


def nejdelsi_serie_bez_preskoceni(prehravani):
    nejlepsi = 0
    aktualni = 0
    for p in prehravani:
        if p["preskoceno"]:
            aktualni = 0
        else:
            aktualni += 1
            nejlepsi = max(nejlepsi, aktualni)
    return nejlepsi


def vypis_prehled(prehravani):
    print("=== Tvuj rok v hudbe ===")
    print("Celkem prehrani:        ", pocet_prehrani(prehravani))
    print("Unikatnich interpretu:  ", len(unikatni_interpreti(prehravani)))
    print("Preskoceno (%):         ", round(procento_preskocenych(prehravani), 1))
    print("Nejdelsi serie bez skipu:", nejdelsi_serie_bez_preskoceni(prehravani))
    print("Top 5 interpretu:")
    pocty = pocty_podle_interpreta(prehravani)
    for poradi, (interpret, kolik) in enumerate(top_interpreti(pocty), start=1):
        print("  ", poradi, ".", interpret, "-", kolik, "x")


if __name__ == "__main__":
    vypis_prehled(prehravani)
```
