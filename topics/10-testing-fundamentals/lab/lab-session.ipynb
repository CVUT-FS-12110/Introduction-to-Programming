## Debrief (5 min)

### Co jsme to vlastně dělali

Technika z dnešního cvičení má jméno: **mutační testování** (mutation
testing). Nástroje jako `mutmut` dělají přesně to, co vy dnes ručně:
automaticky vytvářejí „evil" mutace kódu a měří, kolik z nich vaše testy
zabijí. Sada, kterou mutanti procházejí, je slabá — přesně jako ta naše
v kole 1.

### Otázky k zamyšlení

1. Jaká díra v testech tě dnes nejvíc překvapila?
