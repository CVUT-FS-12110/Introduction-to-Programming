# Řešení: Volební algoritmy

```
# Datová sada
volby_2021 = {"SPOLU": 1493905, "ANO": 1458140, "Piráti+STAN": 839776, "SPD": 513910}
CELKEM_MANDATU = 200
```

## Úkol 1: D'Hondtova metoda

```
def dhondt(hlasy, pocet_mandatu):
    mandaty = {strana: 0 for strana in hlasy.keys()}
    
    for _ in range(pocet_mandatu):
        nejvyssi_kvocient = -1
        vitez = None
        
        for strana, pocet_hlasu in hlasy.items():
            kvocient = pocet_hlasu / (mandaty[strana] + 1)
            
            if kvocient > nejvyssi_kvocient:
                nejvyssi_kvocient = kvocient
                vitez = strana
                
        mandaty[vitez] += 1
        
    return mandaty

print("D'Hondt:", dhondt(volby_2021, CELKEM_MANDATU))
```

## Úkol 2: Sainte-Laguë metoda

```
def sainte_lague(hlasy, pocet_mandatu):
    mandaty = {strana: 0 for strana in hlasy.keys()}
    
    for _ in range(pocet_mandatu):
        nejvyssi_kvocient = -1
        vitez = None
        
        for strana, pocet_hlasu in hlasy.items():
            # Dělitel je lichá řada: 1, 3, 5, 7 = 2 * n + 1
            kvocient = pocet_hlasu / (2 * mandaty[strana] + 1)
            
            if kvocient > nejvyssi_kvocient:
                nejvyssi_kvocient = kvocient
                vitez = strana
                
        mandaty[vitez] += 1
        
    return mandaty

print("Sainte-Laguë:", sainte_lague(volby_2021, CELKEM_MANDATU))
```

## Úkol 3: Bez klauzule

```
volby_2021_vsechny = {
    "SPOLU": 1493905, "ANO": 1458140, "Piráti+STAN": 839776, "SPD": 513910, 
    "Přísaha": 251562, "ČSSD": 250397, "KSČM": 193817, "Trikolora": 149720, "Zelení": 53583
}

bez_klauzule = dhondt(volby_2021_vsechny, CELKEM_MANDATU)

print("Kdo by se dostal do sněmovny bez 5% bariéry:")
for strana, m in sorted(bez_klauzule.items(), key=lambda x:-x[1]):
    if m > 0:
        print(f"{strana}: {m} mandátů")
```
